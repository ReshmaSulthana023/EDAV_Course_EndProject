
Q1: Calculate disease incidence mean (BL-3)

Here the .mean() function calculates the mean of the incidence per 100k of all states from the data.
It helps to analyse the national-level disease incidences.

```python

mean_incidence=df["incidence_per_100k"].mean()
print("Disease incidence mean:\n",mean_incidence)
```


Q2: Filter data for specific diseases (BL-3)

Here .isin() function checks whether the disease is of malaria or dengue or measles and filters the data accordingly.

```python
Filtered_data=df[df['disease'].isin(["Malaria","Dengue","Measles"])]
print("Filtered data of dengue,malaria,measles:\n",Filtered_data)
```


Q3: Impute missing incidence data (BL-4)

There might be empty places(NaN) in the data.So the .fillna() function comes into the picture to handle them.
We can fill these empty places with a particular value/mean of the column/median/mode,etc...
This standardize the data for accurate analysis.

```python

print("empty incidence before imputing:")
print(df["incidence_per_100k"])

print("\nFilled incidence after imputing:\n")
print(df["incidence_per_100k"].fillna(mean_incidence))
```


Q4: Group and summarize by disease and state (BL-4)

groupby() groups the dataset by disease and state.
agg() applies multiple summary functions like count,mean,median,sum,max and min.
reset_index() restores the grouped result as a DataFrame.
This produces a detailed summary showing which disease dominate in which states.

```python

grouped_data=df.groupby(["disease","state"])

summarizing=grouped_data["incidence_per_100k"].agg(
    count='count',
    mean='mean',
    median='median',
    sum='sum',
    max='max',
    min='min').reset_index()
print("\n\nSummary of the data from incidence_per_100k column:\n",summarizing)

print("\n\nGrouped data by disease and states:\n")
for name,group in grouped_data:
    print("Group",name)
    print(group.head(5))
    print("-"*40)
```

Q5: Plot incidence trends by region (BL-5)

The variable x stores the values of incidence per 100k column and y stores the names of different regions in the dataset.
plt.title() gives the title for graph.We can give label for x-axis and y-axis  using .xlabel() and .ylabel().The bar graph can be printed by using .show() function.

```python
x=df["incidence_per_100k"]
y=df["region"]
plt.bar(x,y,color='green')
plt.title("Incidence trends by region")
plt.xlabel("Incidence")
plt.ylabel("Region")
plt.legend(title='Region(Average incidence per 100k)')
plt.show()
```
